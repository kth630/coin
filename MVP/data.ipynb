# 업비트 api요청

```python
import pandas as pd
import requests
import pandas_ta as ta
```

```python
to = '2025-09-17T09:00'
url = "https://api.upbit.com/v1/candles/days"
btc = [] # btc.extend(btc_json)
headers = {"accept": "application/json"}

while to[:4] != '2019':
    params = {
    'market' : 'KRW-BTC',
    'to' : to,
    'count': 200,
    'converting_price_unit':'KRW'
    }
    response = requests.get(url, headers=headers,params=params)
    btc_json = response.json()
    btc.extend(btc_json)
    to = btc_json[-1]['candle_date_time_kst']
    print(f"{btc_json[0]['candle_date_time_kst'][:10]}~{btc_json[-1]['candle_date_time_kst'][:10]} ok")
print('Done!')
```

```python
df = pd.DataFrame(btc)[[ 'candle_date_time_kst', 'opening_price', 'high_price', 'low_price', 'trade_price','candle_acc_trade_volume']]
df = df.rename(columns={ 'candle_date_time_kst':'date', 'opening_price':'open', 'high_price':'high', 'low_price':'low', 'trade_price':'close','candle_acc_trade_volume':'volume'})
df
df['date'] = pd.to_datetime(df['date'])
df.sort_values('date',inplace=True)
# df.set_index('date')
df.drop_duplicates(ignore_index=True,inplace=True)

```

```python
df.to_csv('./coins/btc(krw)_2020-2025.csv',index=False)
```
